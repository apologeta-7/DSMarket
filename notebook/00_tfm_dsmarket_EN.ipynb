## 1. Project overview

DSMarket, formerly known as TradiStores, is a small supermarket chain in the United States
that is in an early stage of digital transformation.

Within that context, this Master's Final Project addresses a use case focused on demand forecasting
and its potential application to in-store stock replenishment.

The project is framed as a role-play exercise in which we take on the role of Nicole,
a senior data scientist joining the company to drive priority analytics initiatives
for the finance and operations areas.

The company needs to evolve from manual processes and rudimentary forecasts
toward more structured decision-making supported by data and predictive models.

<a id="preguntas"></a>

## 2. Key project questions

This Master's Final Project is structured around four main questions:

1. **Is it still valid to forecast at store × product level and then aggregate to higher levels?**
2. **How can we build a reasonable and defensible 28-day demand forecast?**
3. **How can that forecast be connected to an operational stock replenishment logic?**
4. **How can we define a realistic productization path through modular code and an API?**

### General approach

The project combines three complementary layers:

- **business interpretation**, to understand the problem and its operational implications,
- **analysis and modeling**, to build a reasonable predictive solution,
- **technical proposal**, to show how this solution could evolve toward more operational use.

## 3. TFM structure

The project has been organized into several notebooks, each with a clear function within the overall work:

| Notebook | Main purpose |
|----------|--------------|
| `00_TFM_DSMarket.ipynb` | Cover, table of contents and project reading guide |
| `01_eda-business.ipynb` | Exploratory analysis and business interpretation |
| `02_clustering.ipynb` | Product segmentation and contextual analysis |
| `03_forecasting-final.ipynb` | Forecasting model construction and evaluation |
| `04_stock.ipynb` | Proposal linking forecasting and replenishment |
| `05_pipeline_api.ipynb` | Technical demonstration of modular pipeline and API |

### Reading logic

The project sequence follows this logic:

1. **Understand the business and demand behavior**
2. **Explore whether differentiated patterns or segments exist**
3. **Build a forecasting model**
4. **Translate the forecast into an operational proposal**
5. **Show a possible productization path**

This structure aims to keep the responsibilities of each notebook clearly separated
and avoid mixing analysis, modeling and technical proposal in a single document.

## 4. Technical architecture and environments

The project was developed across several execution environments,
using the most appropriate one in each case according to the computational cost of the task.

This decision is not due to a lack of consistency, but to a practical choice
focused on reproducibility and technical feasibility.

| Notebook | Main environment | Reason |
|----------|------------------|--------|
| `01_eda-business.ipynb` | Colab / Kaggle | Exploratory analysis with a manageable dataset |
| `02_clustering.ipynb` | Local / Kaggle | Work on already aggregated data |
| `03_forecasting-final.ipynb` | Kaggle | Higher memory requirements for the modeling panel |
| `04_stock.ipynb` | Any environment | Proposal document, no heavy workload |
| `05_pipeline_api.ipynb` | Local | Artifact consumption and technical demonstration |

### Modular project layer

In addition to the notebooks, the project includes a reusable code layer in `src/`,
designed to separate business logic, preprocessing, utilities and API service.

At a simplified level, the project structure is as follows:

```text
TFM_DSMarket/
│
├── 00_TFM_DSMarket.ipynb
├── 01_eda-business.ipynb
├── 02_clustering.ipynb
├── 03_forecasting-final.ipynb
├── 04_stock.ipynb
├── 05_pipeline_api.ipynb
│
├── data/
├── models/
├── outputs/
└── src/
    ├── config.py
    ├── paths.py
    ├── preprocessing.py
    ├── forecasting_utils.py
    ├── plotting.py
    └── api/
```

## 5. Scope, reproducibility and reading criteria

This notebook 00 is not intended to replace the technical detail of the rest of the Master's Final Project.
Its role is to introduce the project, organize the reading flow and define the scope of the work.

### Real project scope

This Master's Final Project does not aim to build a complete end-to-end production platform.
The objective is to develop a coherent analytics solution covering:

- understanding of the business problem,
- construction of a demand forecasting model,
- proposed use for stock replenishment,
- and a technical demonstration of a possible productization path.

### Reproducibility

Because the project was developed across different environments,
the following reading criteria are established:

- each notebook acts as the main evidence for its own task,
- the forecasting notebook concentrates the model validation,
- the pipeline/API notebook does not redefine analytical results,
  but demonstrates a possible form of technical consumption,
- small differences resulting from reruns should not change
  the overall interpretation of the project, unless they materially alter
  the business conclusions.

### How this work should be read

The recommended reading order is:

- first, understand the business context and story,
- then, review the exploratory analysis,
- next, evaluate forecasting as the technical core,
- and finally, interpret the replenishment proposal and modular technical layer
  as a reasonable evolution of the use case.
