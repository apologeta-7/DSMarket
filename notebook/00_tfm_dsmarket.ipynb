## 1. Presentación del proyecto

DSMarket, anteriormente conocida como TradiStores, es una pequeña cadena de supermercados en Estados Unidos
que se encuentra en una fase temprana de transformación digital.

Dentro de ese contexto, este Trabajo Final de Máster aborda un caso de uso centrado en la predicción de demanda
y su posible aplicación a la reposición de stock en tiendas.

El proyecto se plantea como un ejercicio de role play en el que asumimos el papel de Nicole,
científica de datos sénior incorporada a la compañía para impulsar iniciativas analíticas prioritarias
para el área financiera y operativa.

La empresa necesita evolucionar desde procesos manuales y predicciones rudimentarias
hacia una toma de decisiones más estructurada, apoyada en datos y modelos predictivos.

<a id="preguntas"></a>

## 2. Preguntas clave del trabajo

Este TFM se articula alrededor de cuatro preguntas principales:

1. **¿Sigue siendo válido predecir a nivel tienda × producto y agregar después a niveles superiores?**
2. **¿Cómo construir un forecasting de demanda con horizonte de 28 días de forma razonable y defendible?**
3. **¿Cómo conectar ese forecasting con una lógica operativa de reposición de stock?**
4. **¿Cómo plantear una vía realista de productivización mediante código modular y una API?**

### Enfoque general del trabajo

El proyecto combina tres capas complementarias:

- **lectura de negocio**, para entender el problema y sus implicaciones operativas,
- **análisis y modelado**, para construir una solución predictiva razonable,
- **planteamiento técnico**, para mostrar cómo esta solución podría evolucionar hacia un uso más operativo.

Esta estructura busca separar bien las responsabilidades de cada notebook

## 4. Arquitectura técnica y entornos

El proyecto se ha desarrollado en varios entornos de ejecución,
utilizando en cada caso el más adecuado según el coste computacional de la tarea.

Esta decisión no responde a una falta de coherencia, sino a una elección práctica
orientada a la reproducibilidad y a la viabilidad técnica del trabajo.

| Notebook | Entorno principal | Motivo |
|----------|-------------------|--------|
| `01_EDA_business_story.ipynb` | Colab / Kaggle | Análisis exploratorio con dataset manejable |
| `02_clustering.ipynb` | Local / Kaggle | Trabajo sobre datos ya agregados |
| `03_forecasting.ipynb` | Kaggle | Mayor necesidad de memoria para el panel de modelado |
| `04_stock_replenishment.ipynb` | Cualquier entorno | Documento de propuesta, sin carga pesada |
| `05_pipeline_api.ipynb` | Local | Consumo de artefactos y demostración técnica |

### Capa modular del proyecto

Además de los notebooks, el trabajo incluye una capa de código reutilizable en `src/`,
pensada para separar lógica de negocio, preprocesado, utilidades y servicio de API.
